```python
import pandas as pd
```

```python
df = pd.read_csv('student_performance.csv')
```

```python
print("Missing values:", df.isna().sum())
```

```python
print("Duplicate rows:", df.duplicated())
```

```python
df.rename(columns={"Student_ID": "Student_ID",
    "Python": "Python_Score",
    "Mathematics": "Mathematics_Score",
    "Statistics": "Statistics_Score",
    "Machine_Learning": "ML_Score",
    "Attendance": "Attendance_Percentage"}, inplace=True)
```

```python
df = df.astype({
    "Python_Score": "int",
    "Mathematics_Score": "int",
    "Statistics_Score": "int",
    "ML_Score": "int",
    "Attendance_Percentage": "float"
})
```

```python
print("Data Types:", df.dtypes)
```

```python
df.head()
```

```python
for id in df['Student_ID']:
    df["Average_Score"] = (df["Python_Score"] + df["Mathematics_Score"] + df["Statistics_Score"] + df["ML_Score"]) / 4
```

```python
performance = []

for score in df["Average_Score"]:
    if score >= 90:
        performance.append("Excellent")
    elif score >= 80:
        performance.append("Good")
    elif score >= 70:
        performance.append("Average")
    else:
        performance.append("Needs Improvement")

df["Performance"] = performance
```

```python
df.to_csv('cleaned_student_performance.csv', index=False)
```
