```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns   
```

```python
df = pd.read_csv('../cleaned_student_performance.csv')
```

```python
df.columns
```

```python
sns.barplot(x="Student_Name", y="Average_Score", data=df)
plt.xticks(rotation=45, fontsize=8) 
plt.title("Average Score of Each Student", fontsize=14, fontweight="bold")
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df['Attendance_Percentage'], bins=10)
plt.title("Distribution of Students' Attendance Percentages", fontsize=14, fontweight="bold")
```

```python
sns.countplot(x="Degree_Program", data=df, palette="Set2")
plt.title("Number of Students Enrolled in Each Degree Program", fontsize=14, fontweight="bold")
plt.xticks(rotation=45)
```

```python
plt.figure(figsize=(8,5))

sns.scatterplot( data=df, x="Python_Score", y="ML_Score")

plt.title("Python vs Machine Learning Marks", fontsize=14, fontweight="bold")
plt.show()
```

```python
performance_count = df["Performance"].value_counts()

plt.figure(figsize=(7,7))

plt.pie(
    performance_count,
    labels=performance_count.index,
    autopct="%1.1f%%"
)

plt.title("Performance Categories", fontsize=14, fontweight="bold")
plt.show()
```
