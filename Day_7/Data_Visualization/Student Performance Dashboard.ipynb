```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#
# Load Dataset
df = pd.read_csv("../cleaned_student_performance.csv")

sns.set_style("whitegrid")

print("=" * 60)
print("        STUDENT PERFORMANCE DASHBOARD")
print("=" * 60)
```

```python
# Total Students

total_students = len(df)

print(f"\nTotal Students : {total_students}")

# Average Score Per Subject

subjects = [
    "Mathematics_Score",
    "Statistics_Score",
    "Python_Score",
    "ML_Score"
]

subject_average = df[subjects].mean()

print("\nAverage Score Per Subject")
print(subject_average)

# Top 5 Students

print("\nTop 5 Performing Students")

top5_students = df.nlargest(5, "Average_Score")[
    ["Student_ID", "Student_Name", "Average_Score", "Performance"]
]

print(top5_students)

# Students Needing Improvement

print("\nStudents Needing Improvement")

needs_improvement = df[
    df["Performance"] == "Needs Improvement"
][["Student_ID", "Student_Name", "Average_Score"]]

print(needs_improvement)

# Subject With Highest Average

highest_subject = subject_average.idxmax()
highest_average = subject_average.max()

print("\nSubject With Highest Average")
print(f"{highest_subject} ({highest_average:.2f})")
```

```python

# VISUALIZATIONS

# Bar Chart

plt.figure(figsize=(8,5))

sns.barplot(
    x=subject_average.index,
    y=subject_average.values
)

plt.title("Average Score Per Subject")
plt.xlabel("Subjects")
plt.ylabel("Average Score")

plt.tight_layout()
plt.show()

# Top 5 Students

plt.figure(figsize=(8,5))

sns.barplot(
    data=top5_students,
    x="Student_Name",
    y="Average_Score"
)

plt.title("Top 5 Performing Students")
plt.xlabel("Students")
plt.ylabel("Average Score")

plt.tight_layout()
plt.show()

# Performance Categories

performance_count = df["Performance"].value_counts()

plt.figure(figsize=(6,6))

plt.pie(
    performance_count,
    labels=performance_count.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Performance Categories")

plt.tight_layout()
plt.show()

# Average Score Distribution

plt.figure(figsize=(8,5))

sns.histplot(
    df["Average_Score"],
    bins=10,
    kde=True
)

plt.title("Average Score Distribution")
plt.xlabel("Average Score")

plt.tight_layout()
plt.show()

# Python vs Machine Learning

plt.figure(figsize=(8,5))

sns.scatterplot(
    data=df,
    x="Python_Score",
    y="ML_Score",
    hue="Performance"
)

plt.title("Python vs Machine Learning")

plt.tight_layout()
plt.show()

print("\nDashboard Analysis Completed Successfully.")
```
